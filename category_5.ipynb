```python
import heapq

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': ['G'],
    'G': []
}

h = {'A': 6, 'B': 4, 'C': 2, 'D': 5, 'E': 3, 'F': 1, 'G': 0}

def greedy_bfs(start, goal):
    pq = [(h[start], start)]
    visited = set()

    while pq:
        _, node = heapq.heappop(pq)

        if node == goal:
            print("Goal reached:", node)
            return

        if node not in visited:
            print(node, end=" ")
            visited.add(node)

            for neighbor in graph[node]:
                if neighbor not in visited:
                    heapq.heappush(pq, (h[neighbor], neighbor))

    print("Goal not found")

greedy_bfs('A', 'G')
```

```python
import heapq

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 2)],
    'D': [('G', 5)],
    'E': [('G', 1)],
    'F': [('G', 3)],
    'G': []
}

h = {'A': 7, 'B': 6, 'C': 4, 'D': 4, 'E': 1, 'F': 3, 'G': 0}

def astar(start, goal):
    pq = [(h[start], 0, start, [start])]
    visited = {}

    while pq:
        f, g, node, path = heapq.heappop(pq)

        if node == goal:
            print("Path:", path)
            print("Cost:", g)
            return

        if node in visited and visited[node] <= g:
            continue

        visited[node] = g

        for neighbor, cost in graph[node]:
            new_g = g + cost
            heapq.heappush(
                pq, (new_g + h[neighbor], new_g,
                     neighbor, path + [neighbor])
            )

astar('A', 'G')
```

```python
graph = {
    'A': [[('B', 1), ('C', 1)], [('D', 4)]],
    'B': [[('E', 2)]],
    'C': [[('F', 2)]],
    'D': [],
    'E': [],
    'F': []
}

h = {'A': 5, 'B': 2, 'C': 2, 'D': 1, 'E': 0, 'F': 0}

def ao_star(node):
    if not graph[node]:
        return h[node]

    costs = []

    for option in graph[node]:
        cost = 0

        for child, edge_cost in option:
            cost += edge_cost + ao_star(child)

        costs.append(cost)

    h[node] = min(costs)
    return h[node]

cost = ao_star('A')
print("Minimum cost:", cost)
print("Updated heuristic values:", h)
```

```python
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2)],
    'C': [('G', 3)],
    'D': [('G', 2)],
    'G': []
}

h = {'A': 4, 'B': 3, 'C': 3, 'D': 2, 'G': 0}

def ida_star(start, goal):
    bound = h[start]
    path = [start]

    def search(node, g, bound):
        f = g + h[node]

        if f > bound:
            return f

        if node == goal:
            return True

        minimum = float('inf')

        for neighbor, cost in graph[node]:
            if neighbor not in path:
                path.append(neighbor)

                result = search(neighbor, g + cost, bound)

                if result is True:
                    return True

                minimum = min(minimum, result)
                path.pop()

        return minimum

    while True:
        result = search(start, 0, bound)

        if result is True:
            print("Path:", path)
            return

        if result == float('inf'):
            print("Goal not found")
            return

        bound = result

ida_star('A', 'G')
```
