```python
#1. Breadth First Search

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}


def bfs(start):
    queue = [start]
    visited = []

    while queue:
        current = queue.pop(0)

        if current not in visited:
            visited.append(current)

            for neighbor in graph[current]:
                if neighbor not in visited:
                    queue.append(neighbor)

    return visited


# Perform BFS
result = bfs("A")

print("Graph:")
print("""
        A
       / \\
      B   C
     / \\   \\
    D   E   F
""")

print("BFS Traversal:")
print(" -> ".join(result))
```

```python
#2. Depth First Search

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}


def dfs(node, visited):
    visited.append(node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(neighbor, visited)


# Start DFS
visited = []
dfs("A", visited)

print("Graph:")
print("""
        A
       / \\
      B   C
     / \\   \\
    D   E   F
""")

print("DFS Traversal:")
print(" -> ".join(visited))
```

```python
#3. Depth Limited Search

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}


def dls(node, depth, limit, visited):

    visited.append(node)

    # Stop when depth limit is reached
    if depth == limit:
        return

    for neighbor in graph[node]:
        if neighbor not in visited:
            dls(neighbor, depth + 1, limit, visited)


# Get a valid depth limit from user
while True:
    limit = int(input("Enter depth limit: "))

    if limit <= 2:
        break

    print("Depth limit is only up to 2 for this graph. Please enter again.")


visited = []

# Start DLS from A
dls("A", 0, limit, visited)


print("\nGraph:")
print("""
        A
       / \\
      B   C
     / \\   \\
    D   E   F
""")

print("Depth Limit:", limit)
print("DLS Traversal:")
print(" -> ".join(visited))
```

```python
#4. Bidirectional Search

graph = {
    "A": ["B"],
    "B": ["A", "C", "D"],
    "C": ["B", "E"],
    "D": ["B", "G"],
    "E": ["C", "F"],
    "F": ["E"],
    "G": ["D"]
}


def bidirectional_search(start, goal):

    front_start = [start]
    front_goal = [goal]

    visited_start = [start]
    visited_goal = [goal]

    while front_start and front_goal:

        # Search from the start side
        current = front_start.pop(0)

        for neighbor in graph[current]:

            if neighbor not in visited_start:
                visited_start.append(neighbor)

                # Check whether the two searches have met
                if neighbor in visited_goal:
                    return neighbor

                front_start.append(neighbor)

        # Search from the goal side
        current = front_goal.pop(0)

        for neighbor in graph[current]:

            if neighbor not in visited_goal:
                visited_goal.append(neighbor)

                # Check whether the two searches have met
                if neighbor in visited_start:
                    return neighbor

                front_goal.append(neighbor)

    return None


start = input("Enter start node: ").upper()
goal = input("Enter goal node: ").upper()

meeting_point = bidirectional_search(start, goal)


print("\nGraph:")
print("""
A --- B --- D --- G
      |
      C --- E --- F
""")


if meeting_point:
    print("Searches meet at:", meeting_point)
    print("Bidirectional Search successful!")
else:
    print("No meeting point found.")
```
